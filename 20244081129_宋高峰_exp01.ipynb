## 实验 2：函数、条件、循环与递归

### 2.1 分数等级

90、80、60 分为等级分界；超出 0～100 范围抛出 ValueError。

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在 0～100 范围内")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


for score in [0, 59, 60, 79, 80, 89, 90, 100, -1, 101]:
    try:
        print(f"{score} 分：{score_level(score)}")
    except ValueError as error:
        print(f"{score} 分：捕获 ValueError，{error}")
```

### 2.2 循环求和

分别用 for 和 while 累加 1 到 100，比较结果。

```python
for_total = 0
for number in range(1, 101):
    for_total += number

while_total = 0
number = 1
while number <= 100:
    while_total += number
    number += 1

print("for 求和：", for_total)
print("while 求和：", while_total)
print("结果相等：", for_total == while_total)
assert for_total == while_total == 5050
```

### 2.3 递归与迭代阶乘

递归使用 n! = n × (n−1)!，迭代逐项相乘，均以 0! = 1 为基础。

```python
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    result = 1
    for number in range(1, n + 1):
        result *= number
    return result


for n in [0, 1, 5, 10]:
    recursive = factorial_recursive(n)
    iterative = factorial_iterative(n)
    print(f"n = {n:2d}，递归结果 = {recursive:7d}，迭代结果 = {iterative:7d}，相等：{recursive == iterative}")
    assert recursive == iterative
```

### 2.4 元组返回与可变参数

summary() 返回最小值、最大值和平均值；sum_all() 接收任意数量的参数。

```python
def summary(values):
    return min(values), max(values), sum(values) / len(values)


minimum, maximum, average = summary([85, 58, 92, 76, 60, 89])
print(f"解包结果：最小值={minimum}，最大值={maximum}，平均值={average:.2f}")


def sum_all(*args):
    return sum(args)


print("无参数：", sum_all())
print("单个参数：", sum_all(7))
print("多个参数：", sum_all(1, 2, 3, 4.5))
```

## 实验 3：列表、字典、元组与基础统计

### 3.1 遍历与列表推导式

用 zip 配对、items 遍历、enumerate 编号，再提取及格姓名和成绩平方。

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_dict = dict(zip(names, scores))
print("姓名到成绩的字典：", score_dict)
print("\nitems() 逐项输出：")
for name, score in score_dict.items():
    print(f"{name}: {score}")

print("\nenumerate() 从 1 编号：")
for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{index}. {name}: {score}")

passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print("\n及格学生：", passed_names)
print("成绩平方列表：", squared_scores)
```

### 3.2 基础统计

计算人数、平均分、最高分、最低分及及格率，并格式化输出。

```python
student_count = len(scores)
average_score = sum(scores) / student_count
highest_score = max(scores)
lowest_score = min(scores)
pass_rate = sum(score >= 60 for score in scores) / student_count

print(f"人数：{student_count}")
print(f"平均分：{average_score:.2f}")
print(f"最高分：{highest_score}")
print(f"最低分：{lowest_score}")
print(f"及格率：{pass_rate:.2%}")
```

### 3.3 别名与列表副本

直接赋值共享列表，copy() 创建副本；使用单独的演示列表。

```python
demo_scores = scores.copy()
alias = demo_scores
copied = demo_scores.copy()
print("修改前：", demo_scores, alias, copied)
alias[0] = 100
print("修改 alias[0] 后：")
print("演示原列表：", demo_scores)
print("alias：", alias)
print("copied：", copied)
print("alias 与原列表是同一对象：", alias is demo_scores)
print("copied 与原列表是同一对象：", copied is demo_scores)
print("后续实验的原始 scores：", scores)
assert scores == [85, 58, 92, 76, 60, 89]
```

## 实验 4：文件读写、异常处理与模块

### 4.1 文件读写

使用相对路径创建 exp01_data，按 UTF-8 编码写入并读取模拟数据。

```python
print("当前工作目录：", os.getcwd())
data_dir = os.path.join(".", "exp01_data")
os.makedirs(data_dir, exist_ok=True)
students_path = os.path.join(data_dir, "students.txt")

students_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(students_path, "w", encoding="utf-8") as file:
    file.write(students_text)

with open(students_path, "r", encoding="utf-8") as file:
    print("\n写入并读回的 students.txt：")
    print(file.read(), end="")
```

### 4.2 记录解析

检查字段数量、姓名、整数成绩及成绩范围；无效记录抛出 ValueError。

```python
def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError(f"字段数量错误：应为 2 个，实际为 {len(fields)} 个")
    name, score_text = (field.strip() for field in fields)
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(score_text)
    except ValueError as error:
        raise ValueError(f"成绩不是有效整数：{score_text!r}") from error
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0～100 范围：{score}")
    return {"name": name, "score": score}


print("有效记录：", parse_student("Alice,85"))
for line in ["Grace,abc", "Henry,105", "Incomplete", ",85"]:
    try:
        parse_student(line)
    except ValueError as error:
        print(f"{line!r} -> ValueError：{error}")
```

### 4.3 逐行处理

跳过表头，保存有效记录；捕获无效记录的错误并记录行号，继续处理。

```python
valid_records = []
invalid_records = []
with open(students_path, "r", encoding="utf-8") as file:
    header = next(file).strip()
    print("跳过表头：", header)
    for line_number, line in enumerate(file, start=2):
        try:
            valid_records.append(parse_student(line))
        except ValueError as error:
            invalid_records.append({"line_number": line_number, "reason": str(error)})
            print(f"第 {line_number} 行无效：{error}")

print("\n有效记录：")
for record in valid_records:
    print(record)
print(f"有效记录数：{len(valid_records)}；无效记录数：{len(invalid_records)}")
assert len(valid_records) == 6
assert [record["line_number"] for record in invalid_records] == [8, 9, 10]
```

### 4.4 统计报告与写入模式

计算并保存统计结果；w 覆盖已有内容，a 在末尾追加。

```python
valid_scores = [record["score"] for record in valid_records]
valid_count = len(valid_records)
invalid_count = len(invalid_records)
file_average = sum(valid_scores) / valid_count
file_pass_rate = sum(score >= 60 for score in valid_scores) / valid_count
report_path = os.path.join(data_dir, "report.txt")

with open(report_path, "w", encoding="utf-8") as file:
    file.write("实验 01 成绩统计报告\n")
    file.write(f"有效记录数：{valid_count}\n")
    file.write(f"无效记录数：{invalid_count}\n")
    file.write(f"平均分：{file_average:.2f}\n")
    file.write(f"及格率：{file_pass_rate:.2%}\n")

with open(report_path, "a", encoding="utf-8") as file:
    file.write("\n无效记录明细：\n")
    for record in invalid_records:
        file.write(f"第 {record['line_number']} 行：{record['reason']}\n")

with open(report_path, "r", encoding="utf-8") as file:
    print(file.read())

mode_demo_path = os.path.join(data_dir, "mode_demo.txt")
with open(mode_demo_path, "w", encoding="utf-8") as file:
    file.write("原有内容\n")
with open(mode_demo_path, "w", encoding="utf-8") as file:
    file.write("w 模式的新内容\n")
with open(mode_demo_path, "r", encoding="utf-8") as file:
    print("再次使用 w 后：", repr(file.read()))
with open(mode_demo_path, "a", encoding="utf-8") as file:
    file.write("a 模式追加的内容\n")
with open(mode_demo_path, "r", encoding="utf-8") as file:
    print("使用 a 后：", repr(file.read()))
```

## 实验 5：类、继承与对象复制

### 5.1 Student 类

保存姓名和成绩，实现对象描述与及格判断，为有效记录创建对象。

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


students = [Student(record["name"], record["score"]) for record in valid_records]
for student in students:
    print(f"{student}，及格：{student.is_passed()}")
```

### 5.2 继承

用 super() 初始化父类属性，增加研究方向并重写对象描述。

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (
            f"GraduateStudent(name={self.name}, score={self.score}, "
            f"research_topic={self.research_topic})"
        )


graduate = GraduateStudent("Alice", 85, "数据分析")
print(graduate)
print("继承的及格判断：", graduate.is_passed())
print("属于 Student：", isinstance(graduate, Student))
```

### 5.3 对象复制

直接赋值共享整个对象，浅复制共享嵌套列表，深复制得到独立的嵌套列表。

```python
import copy

original = {"name": "Alice", "scores": [85, 90]}
assigned = original
assigned["scores"][0] = 100
print("直接赋值：")
print("original =", original, "assigned =", assigned)
print("外层同一对象：", assigned is original, "内部列表同一对象：", assigned["scores"] is original["scores"])

original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)
shallow["scores"][0] = 100
print("\n浅复制：")
print("original =", original, "shallow =", shallow)
print("外层同一对象：", shallow is original, "内部列表同一对象：", shallow["scores"] is original["scores"])

original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)
deep["scores"][0] = 100
print("\n深复制：")
print("original =", original, "deep =", deep)
print("外层同一对象：", deep is original, "内部列表同一对象：", deep["scores"] is original["scores"])
assert original["scores"] == [85, 90]
assert deep["scores"] == [100, 90]
```
